```python
df["lifestyle_score"] = df["Sleep_Hours"] + df["Exercise_Hours"] - df["Stress_Level"]
df["risk_score"] = df["BMI"] + df["Blood_Pressure"] + df["Cholesterol"] + df["Glucose_Level"]
df["wellness"] = df["Water_Intake"] + df["PhysicalActivity"] + df["MentalHealth"] - df["Allergies"]-df["MedicalHistory"]
```

```python
x = df.drop("Target",axis=1)
y = df["Target"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x = x.drop(columns=["risk_score"])
```

```python
x_train,x_test,y_train,y_test = train_test_split(
    x,y,random_state=42,test_size=0.2
)
```

```python
from sklearn.ensemble import RandomForestClassifier
ref_ = RandomForestClassifier(oob_score=True, n_jobs=-1)
```

```python
from sklearn.model_selection import GridSearchCV

pram = {
    "max_depth": [3,5,9],
    "min_samples_split":[3,6,8,],
    "n_estimators":[100,200,300,],
    
}
grid = GridSearchCV(
    ref_,pram,cv=5,n_jobs=-1,scoring="recall",
)
grid.fit(x_train,y_train)

print(grid.best_params_)
```

```python
from sklearn.ensemble import RandomForestClassifier
ref_ = RandomForestClassifier(max_depth=9, min_samples_split= 3, n_estimators=300, oob_score=True, n_jobs=-1)
```

```python
ref_.fit(x_train,y_train)

y_trian_pred = ref_.predict(x_train)
y_test_pred = ref_.predict(x_test)

from sklearn.metrics import classification_report,r2_score,confusion_matrix
print("classification_report train :",classification_report(y_train,y_trian_pred))
print("classification_report test :",classification_report(y_test,y_test_pred))
```

```python
print("confusion_matrix train :",confusion_matrix(y_train,y_trian_pred))
print("confusion_matrix test :",confusion_matrix(y_test,y_test_pred))
```

```python
# ROC-AUC Score 
from sklearn.metrics import roc_auc_score

y_prob = ref_.predict_proba(x_test)[:,1]
roc_auc_score(y_test, y_prob)
```

```python
# PR Curve 
from sklearn.metrics import precision_recall_curve

y_prob = ref_.predict_proba(x_test)[:,1]
precision, recall, thresholds = precision_recall_curve(y_test, y_prob)
print(precision,recall,thresholds)
```

```python
# feature importance

import pandas as pd

importance = ref_.feature_importances_

pd.Series(
    importance,
    index=x_train.columns
).sort_values(ascending=False)
```
